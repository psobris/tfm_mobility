```python
# ==============================================================================
# NOTEBOOK: 02_raw_batch
# DESCRIPCIÓN: Convierte fuentes Batch de Landing a RAW Parquet
# ==============================================================================
import os
from tfm_mobility.processors.raw_processor import RawProcessor

print("⏳ [RAW BATCH] Iniciando conversión de fuentes batch a Parquet...")
processor = RawProcessor(spark)

def get_latest_file(base_dir):
    try:
        all_files = []
        full_path = os.path.join("/lakehouse/default", base_dir.lstrip("/"))
        if os.path.exists(full_path):
            for root, _, files in os.walk(full_path):
                for f in files:
                    all_files.append(os.path.join(root, f))
        if all_files:
            latest = max(all_files, key=os.path.getmtime)
            return latest.replace("/lakehouse/default/", "")
    except Exception as e:
        print(f"⚠️ No se encontraron archivos en {base_dir}: {e}")
    return None

# 1. NASA HISTÓRICO (7 días)
nasa_batch_path = locals().get('nasa_batch_path') or get_latest_file("Files/landing/batch/nasa_historical")
if nasa_batch_path:
    print(f"📄 Procesando NASA Batch desde: {nasa_batch_path}")
    processor.landing_to_raw_csv(
        landing_path=nasa_batch_path, 
        raw_path="Files/raw/batch/nasa_historical", 
        is_batch=True
    )
else:
    print("⚠️ No hay ficheros de NASA Histórico en Landing Batch.")

# 2. OSM DATA (Carreteras)
osm_batch_path = locals().get('osm_batch_path') or get_latest_file("Files/landing/batch/osm_roads")
if osm_batch_path:
    print(f"📄 Procesando OSM Red Viaria desde: {osm_batch_path}")
    processor.landing_to_raw_osm(
        landing_path=osm_batch_path, 
        raw_path="Files/raw/batch/osm_roads"
    )
else:
    print("⚠️ No hay ficheros de OSM en Landing Batch.")

print("✅ [RAW BATCH OK] Procesamiento Batch a Parquet finalizado.")
```
