```python
# ==============================================================================
# NOTEBOOK: 01_landing_weather
# ==============================================================================
from tfm_mobility.ingesters.realtime.weather_ingester import WeatherIngester

print("🌤️ [INGESTA] Consultando datos meteorológicos en tiempo real...")

ingester = WeatherIngester()
raw_data = ingester.fetch_data()

# Asegurar que cada elemento de la lista se convierta a JSON/dict si es un Response
clean_data = []
for item in raw_data:
    if hasattr(item, "json") and callable(item.json):
        clean_data.append(item.json())
    else:
        clean_data.append(item)

# Guardar en OneLake Landing y obtener la ruta del fichero guardado
weather_file_path = ingester.save_landing(clean_data)

print(f"✅ Ingesta de Clima finalizada: {weather_file_path}")

# Devolver la ruta al orquestador (00_nrt_pipeline_and_audit)
mssparkutils.notebook.exit(weather_file_path)
```
